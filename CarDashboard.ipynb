```python
from ultralytics import YOLO

# 1. Load a lightweight, pre-trained base model
model = YOLO('yolov8n.pt')

# 2. Train using Ultralytics' pre-built, built-in 'coco8' automotive/object dataset
# This dataset is stored locally inside the library, bypassing the internet error!
results = model.train(data='coco8.yaml', epochs=10, imgsz=640)

print("\n--- AI MODEL TRAINING COMPLETED ---")
```

```python
import os
import zipfile
from google.colab import files

print("Gathering training proof charts and metrics...")

# Check if the training directory exists
train_dir = '/content/runs/detect/train'
zip_name = '/content/training_proof.zip'

if os.path.exists(train_dir):
    # Zip up all diagnostic images, graphs, and accuracy sheets
    with zipfile.ZipFile(zip_name, 'w') as zipf:
        for root, dirs, files_in_dir in os.walk(train_dir):
            for file in files_in_dir:
                # We skip the heavy weights folder to keep the zip file tiny
                if 'weights' not in root:
                    file_path = os.path.join(root, file)
                    zipf.write(file_path, os.path.relpath(file_path, train_dir))

    print("\n--- ZIP FILE CREATED ---")
    print("Downloading 'training_proof.zip' to your desktop now...")

    # Trigger an automatic download straight to your local computer
    files.download(zip_name)
else:
    print("Error: Training folder not found. Make sure your model training cell has finished running completely first!")
```
